# Module 3 — Control Flow

This module covers decision making and repetition in Python: `if`, `for`, `while`, and powerful comprehension syntax.

## Learning Objectives

- Use `if`, `elif`, and `else` to branch logic
- Work with `for` loops and `while` loops
- Apply `break` and `continue` correctly
- Write list comprehensions and conditional expressions

## Conditional branching

Use comparison operators and boolean logic to choose which code runs.

```python
score = 82
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
else:
    grade = 'C'
print('Grade:', grade)
```

## Loops with `for`

Iterate over sequences and ranges to repeat work.

```python
for i in range(1, 6):
    print('Step', i)

words = ['apple', 'banana', 'cherry']
for word in words:
    print(word.upper())
```

## `while` loops

Continue repeating until a condition becomes false.

```python
count = 1
while count <= 5:
    print('Count', count)
    count += 1
```

## `break`, `continue`, and loop control

Use `break` to exit early and `continue` to skip the rest of the iteration.

```python
for n in range(1, 11):
    if n % 2 == 0:
        continue
    if n > 7:
        break
    print(n)
```

## List comprehensions

A concise way to build lists from loops and conditions.

```python
squares = [x*x for x in range(1, 11)]
print(squares)

evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)
```

## Exercise 1 — FizzBuzz

Print numbers 1–20 with Fizz/Buzz rules using a loop and conditional logic.

```python
for n in range(1, 21):
    if n % 15 == 0:
        print('FizzBuzz')
    elif n % 3 == 0:
        print('Fizz')
    elif n % 5 == 0:
        print('Buzz')
    else:
        print(n)
```

## Exercise 2 — Prime Checker

Use a loop to check whether a number is prime and print a friendly result.

```python
n = 29
if n < 2:
    prime = False
else:
    prime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            prime = False
            break
print(n, 'is prime:', prime)
```

## Mini-Project — Grade Categorizer

Build a simple loop that reads a list of scores and prints custom feedback for each student.

```python
scores = [95, 81, 76, 58, 89]
for score in scores:
    if score >= 90:
        feedback = 'Excellent'
    elif score >= 80:
        feedback = 'Good'
    elif score >= 70:
        feedback = 'Room to improve'
    else:
        feedback = 'Needs practice'
    print(f'Score {score}: {feedback}')
```
