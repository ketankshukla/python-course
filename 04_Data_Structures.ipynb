## Lists and common operations

Lists are mutable ordered sequences. Use append, insert, remove, and slicing.

```python
fruits = ['apple', 'banana', 'cherry']
fruits.append('date')
fruits.insert(1, 'blueberry')
print(fruits)
fruits.remove('banana')
print('Sliced:', fruits[1:3])
```

## Tuples: fixed sequences

Tuples are like lists but cannot be changed. They are useful for fixed records and unpacking.

```python
point = (10, 20)
print('x:', point[0], 'y:', point[1])
x, y = point
print(x, y)
# point[0] = 5  # would raise an error
```

## Sets and membership

Sets store unique values and support union/intersection/difference.

```python
colors = {'red', 'blue', 'green'}
colors.add('yellow')
print(colors)
print('Has red?', 'red' in colors)
other = {'blue', 'purple'}
print('Intersection:', colors & other)
print('Union:', colors | other)
```

## Dictionaries for structured data

Dictionaries map keys to values. Access values by key and iterate over items.

```python
student = {'name': 'Mia', 'score': 92, 'passed': True}
print(student['name'])
for key, value in student.items():
    print(key, '=>', value)
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print('Common:', a & b)
print('Only in a:', a - b)
```
