## CSV files

CSV is a common text format for table-like data. Use the `csv` module to read and write it.

```python
import csv
rows = [
    ['name', 'score'],
    ['Ava', '92'],
    ['Ben', '85'],
]
with open('grades.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerows(rows)

with open('grades.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## JSON data

JSON is widely used for structured data exchange. Use `json.loads` and `json.dumps`.

```python
import json
record = {'name': 'Mia', 'age': 27, 'skills': ['python', 'data']}
json_text = json.dumps(record)
print(json_text)
parsed = json.loads(json_text)
print(parsed['skills'])
```

## pandas introduction

pandas makes it easy to load, explore, and summarize tabular data.

```python
import pandas as pd
df = pd.DataFrame([{'name': 'Ava', 'score': 92}, {'name': 'Ben', 'score': 85}, {'name': 'Mia', 'score': 97}])
print(df)
print('Mean score:', df['score'].mean())
```

## Exercise 1 — Read CSV and summarize

Read `grades.csv` and print the average score using Python loops or pandas.

```python
import csv
with open('grades.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    scores = [int(row['score']) for row in reader]
print('Average score:', sum(scores) / len(scores))
```

## Exercise 2 — JSON contact card

Create a dictionary with contact details and save it to `contact.json`. Then read it back.

```python
import json
contact = {'name': 'Leo', 'email': 'leo@example.com', 'phone': '555-0000'}
with open('contact.json', 'w', encoding='utf-8') as f:
    json.dump(contact, f, indent=2)
with open('contact.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)
print(loaded)
```

## Mini-Project — Sales Summary

Load sales data from a CSV, compute totals and averages, and present the results.

```python
sales = [
    {'item': 'book', 'quantity': 4, 'price': 12.5},
    {'item': 'pen', 'quantity': 10, 'price': 1.2},
    {'item': 'notebook', 'quantity': 5, 'price': 3.4},
]

total = sum(entry['quantity'] * entry['price'] for entry in sales)
print('Total sales:', total)
avg = total / len(sales)
print('Average per product:', avg)
```
