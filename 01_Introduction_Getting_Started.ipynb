## Exercise 1 — Area of a Circle (Cell 12)

Write an expression that computes the area of a circle with radius `r = 5` and assign it to `area`. Use `pi = 3.14159`. Print the result.

````python
# Exercise: compute area
r = 5
pi = 3.14159
area = pi * r**2
print('Area:', area)
````

## Explanation (Solution for Exercise 1)

The formula for the area of a circle is $A=\pi r^2$. In Python `**` is the exponent operator. We used `r**2` to square the radius.

## Mini-Project — Number Guessing Game (Guided)

Build a small command-line number guessing game. Steps: (1) generate a secret number, (2) prompt user, (3) give hints, (4) track attempts. In Jupyter, emulate input by assigning values.

````python
# Guided solution (works in terminal as a .py file)
import random

def play_guessing_game():
    secret = random.randint(1, 50)
    attempts = 0
    print('I have chosen a number between 1 and 50.')
    while True:
        attempts += 1
        guess = input('Your guess: ')
        try:
            guess = int(guess)
        except ValueError:
            print('Please type a number.')
            continue
        if guess < secret:
            print('Higher.')
        elif guess > secret:
            print('Lower.')
        else:
            print(f'Correct! You guessed in {attempts} attempts.')
            break

# To run in a notebook without interactive input, call functions directly or simulate input.
````

### Solution 1 — Sum and Average

We use `sum()` and `len()` to compute these values.

````python
numbers = [2, 5, 7, 11]
total = sum(numbers)
average = total / len(numbers)
print('Sum:', total)
print('Average:', average)
````

### Solution 2 — Swap Variables

Python supports tuple unpacking to swap without a temporary variable.

````python
a = 3
b = 8
a, b = b, a
print('a =', a, 'b =', b)
````

### Solution 3 — String Slicing

Find the start index of the substring or use negative indexing.

````python
s = 'PythonLearner'
sub = s[6:]  # characters from index 6 to end
print(sub)
````

### Solution 4 — Count Vowels

Iterate characters and check membership in a vowel set. Use `.lower()` to simplify.

````python
def count_vowels(text):
    vowels = set('aeiou')
    count = 0
    for ch in text.lower():
        if ch in vowels:
            count += 1
    return count

print(count_vowels('Hello World'))  # expected 3
````

### Solution 5 — FizzBuzz (1–20)

Use modulo `%` to check divisibility and check the multiple-of-both case first.

````python
for n in range(1, 21):
    if n % 15 == 0:
        print('FizzBuzz')
    elif n % 3 == 0:
        print('Fizz')
    elif n % 5 == 0:
        print('Buzz')
    else:
        print(n)
````

### Solution 6 — List Comprehension

Build a comprehension with a conditional to filter evens.

````python
squares = [x*x for x in range(1, 11) if x % 2 == 0]
print(squares)
````

### Solution 7 — File Write/Read (Explanation and Example)

Open a file with `open('notes.txt', 'w')`, write lines with `.write()` or `print(..., file=f)`, close (or use `with`), then open with `'r'` and read lines. Example below writes and reads.

````python
lines = ['Line one\n', 'Line two\n', 'Line three\n']
with open('notes.txt', 'w', encoding='utf-8') as f:
    f.writelines(lines)

with open('notes.txt', 'r', encoding='utf-8') as f:
    data = f.read()
print(data)
````

### Solution 8 — Guessing Game Extension

Limit attempts by counting and break after the maximum; reveal the secret if attempts exhausted.

````python
import random

def play_guessing_game_max(attempts_allowed=6):
    secret = random.randint(1, 50)
    attempts = 0
    while attempts < attempts_allowed:
        attempts += 1
        guess = input(f'Attempt {attempts}/{attempts_allowed} - Your guess: ')
        try:
            guess = int(guess)
        except ValueError:
            print('Please enter a valid integer.')
            continue
        if guess < secret:
            print('Higher.')
        elif guess > secret:
            print('Lower.')
        else:
            print(f'Correct! You guessed in {attempts} attempts.')
            return True
    print(f'Sorry — the secret was {secret}.')
    return False
````

# Factorial (iterative)
def factorial(n):
    if n < 0:
        raise ValueError('n must be >= 0')
    result = 1
    for i in range(2, n+1):
        result *= i
    return result

# Prime check
import math

def is_prime(n):
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2
    return True

# Palindrome (ignoring non-alphanumerics)
import re

def is_palindrome(text):
    cleaned = re.sub(r'[^a-z0-9]', '', text.lower())
    return cleaned == cleaned[::-1]
