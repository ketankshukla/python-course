# Module 6 — File I/O & Error Handling

This module teaches reading and writing files, using context managers, and handling errors safely.

## Learning Objectives

- Open files for reading and writing
- Use `with` to manage file resources
- Handle exceptions with `try/except`
- Raise errors intentionally when needed

## Writing to files

Open a file in write mode and save text.

```python
with open('notes.txt', 'w', encoding='utf-8') as f:
    f.write('Line one\n')
    f.write('Line two\n')
print('Wrote notes.txt')
```

## Reading from files

Open the same file in read mode and process its contents.

```python
with open('notes.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()
print('Read', len(lines), 'lines')
for line in lines:
    print(line.strip())
```

## Handling errors

Use `try` and `except` to catch runtime problems and keep your program from crashing.

```python
try:
    with open('missing.txt', 'r', encoding='utf-8') as f:
        text = f.read()
except FileNotFoundError:
    print('File not found, please create it.')
```

## `else` and `finally`

Use `else` after a successful try and `finally` for cleanup no matter what.

```python
try:
    number = int('42')
except ValueError:
    print('Not a number')
else:
    print('Parsed number:', number)
finally:
    print('Done.')
```

## Exercise 1 — Word frequency file writer

Read a line of text, count words, and write the result to `frequency.txt`.

```python
text = 'hello hello world'
words = text.split()
counts = {word: words.count(word) for word in set(words)}
with open('frequency.txt', 'w', encoding='utf-8') as f:
    for word, count in counts.items():
        f.write(f'{word}: {count}\n')
print('Wrote frequency.txt')
```

## Exercise 2 — Safe file reader

Write code that attempts to open `data.txt`. If it does not exist, print a helpful message.

```python
try:
    with open('data.txt', 'r', encoding='utf-8') as f:
        print(f.read())
except FileNotFoundError:
    print('Create data.txt in the current folder and run again.')
```

## Mini-Project — Simple Log Analyzer

Write a tiny script that reads `notes.txt`, counts lines, and reports summary info. Use exception handling for missing files.

```python
try:
    with open('notes.txt', 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f]
        print('Total lines:', len(lines))
        print('First line:', lines[0] if lines else '(empty)')
except FileNotFoundError:
    print('notes.txt not found. Please create it first.')
```
