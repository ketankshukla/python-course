## Writing tests with pytest

Create a file that starts with `test_` and write functions that use plain `assert` statements.

```python
def add(a, b):
    return a + b

def test_add():
    assert add(1, 2) == 3

print('Create this in a file named test_example.py and run pytest.')
```

## `unittest` basics

Python's standard library also provides `unittest`, which is class-based.

```python
import unittest

class TestMath(unittest.TestCase):
    def test_sum(self):
        self.assertEqual(2 + 3, 5)

if __name__ == '__main__':
    unittest.main()
```

## Debugging techniques

Start with good output, then inspect tracebacks and use a debugger when needed.

```python
def divide(a, b):
    return a / b

try:
    divide(10, 0)
except ZeroDivisionError as e:
    print('Error:', e)
    import traceback
    traceback.print_exc()
```

## Logging

Use the logging module instead of print statements for real applications.

```python
import logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)
logger.info('Program started')
logger.warning('This is a warning')
```

## Exercise 2 — Catch exceptions

Write code that divides two numbers and catches invalid input or division by zero.

```python
def safe_divide(a, b):
    try:
        return a / b
    except (TypeError, ZeroDivisionError):
        return None

print(safe_divide(10, 0))
print(safe_divide('a', 2))
```
