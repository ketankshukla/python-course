## Classes and instances

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        return f'{self.name} says woof!'
```

## Inheritance

Create subclasses that extend or customize behavior from a parent class.

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def drive(self):
        return 'Driving'

class Car(Vehicle):
    def drive(self):
        return 'Driving a car'

my_car = Car('Toyota', 'Corolla')
print(my_car.make, my_car.model)
print(my_car.drive())
```

## Properties and encapsulation

Use `@property` to expose computed values while hiding internal logic.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height

rect = Rectangle(3, 5)
print('Area:', rect.area)
```

## Exercise 1 — Create a class

Create a `Student` class with `name` and `grade`, plus a method that returns a summary string.

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def summary(self):
        return f'{self.name} has grade {self.grade}'

print(Student('Nina', 'A').summary())
```

## Exercise 2 — Subclass behavior

Write a subclass `Teacher` from a `Person` base class and override a greeting method.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f'Hello, I am {self.name}'

class Teacher(Person):
    def greet(self):
        return f'Hello, I am {self.name}, and I teach.'

print(Teacher('Lena').greet())
```

## Mini-Project — Bank Account

Implement a `BankAccount` class with deposit, withdraw, and balance methods plus basic validation.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError('Insufficient funds')
        self.balance -= amount
        return self.balance

account = BankAccount('Dana', 100)
print(account.deposit(50))
print(account.withdraw(30))
```
