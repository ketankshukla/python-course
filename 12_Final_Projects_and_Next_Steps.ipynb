## Final Project Idea 1 — Personal Finance Tracker

Build a small program that reads transactions from a CSV or JSON file, summarizes spending categories, and writes a report.

```python
transactions = [
    {'category': 'food', 'amount': 12.5},
    {'category': 'food', 'amount': 8.0},
    {'category': 'transport', 'amount': 15.75},
]
summary = {}
for tx in transactions:
    category = tx['category']
    summary[category] = summary.get(category, 0) + tx['amount']
print('Summary:', summary)
```

## Final Project Idea 2 — Quiz App

Create a quiz application that asks questions, collects answers, and keeps score. Use functions, loops, and maybe JSON for questions.

```python
questions = [
    {'q': 'What is 2+2?', 'a': '4'},
    {'q': 'What language are we learning?', 'a': 'Python'},
]
score = 0
for item in questions:
    answer = item['a']  # simulate user input
    if answer.lower() == item['a'].lower():
        score += 1
print('Score:', score, '/', len(questions))
```
