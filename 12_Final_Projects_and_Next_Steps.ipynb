## Project 1 — Personal Finance Tracker (Complete Step-by-Step)

### Overview
Build a command-line app that tracks spending by category, reads/writes CSV files, and generates reports. This reinforces: file I/O, data structures, functions, and error handling.

### Step 1: Define the data structure
Think of each transaction as a dictionary with `date`, `category`, `description`, and `amount`.

### Step 2: Create functions for core operations
- `add_transaction(transactions, category, amount, description)` — add a new transaction
- `load_transactions(filename)` — read from CSV or JSON
- `save_transactions(transactions, filename)` — write to file
- `get_summary(transactions)` — sum by category
- `print_report(transactions)` — display friendly output

### Step 3: Build the main loop
Create a menu: add, view, save, quit. Let the user interact until they exit.

### Step 4: Test and refine
Test edge cases: empty transactions, invalid input, missing files.

### Example implementation

```python
import csv
from pathlib import Path

def add_transaction(transactions, category, amount, description=''):
    """Add a transaction to the list."""
    transactions.append({
        'category': category,
        'amount': amount,
        'description': description
    })
    return transactions

def get_summary(transactions):
    """Summarize spending by category."""
    summary = {}
    for tx in transactions:
        cat = tx['category']
        summary[cat] = summary.get(cat, 0) + tx['amount']
    return summary

def print_report(transactions):
    """Print a friendly spending report."""
    if not transactions:
        print('No transactions yet.')
        return
    summary = get_summary(transactions)
    print('\n--- Spending Report ---')
    total = 0
    for category, amount in sorted(summary.items()):
        print(f'{category.capitalize()}: ${amount:.2f}')
        total += amount
    print(f'Total: ${total:.2f}')
    print()

def save_to_csv(transactions, filename='transactions.csv'):
    """Save transactions to a CSV file."""
    try:
        with open(filename, 'w', newline='', encoding='utf-8') as f:
            if not transactions:
                return
            fieldnames = transactions[0].keys()
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(transactions)
        print(f'Saved to {filename}')
    except Exception as e:
        print(f'Error saving: {e}')

def load_from_csv(filename='transactions.csv'):
    """Load transactions from a CSV file."""
    try:
        transactions = []
        if Path(filename).exists():
            with open(filename, 'r', encoding='utf-8') as f:
                reader = csv.DictReader(f)
                for row in reader:
                    row['amount'] = float(row['amount'])
                    transactions.append(row)
        return transactions
    except Exception as e:
        print(f'Error loading: {e}')
        return []

# Main program
transactions = []

while True:
    print('\n1. Add transaction')
    print('2. View report')
    print('3. Save and quit')
    choice = input('Choose: ').strip()
    
    if choice == '1':
        category = input('Category: ').strip()
        try:
            amount = float(input('Amount: ').strip())
            add_transaction(transactions, category, amount)
            print('Transaction added.')
        except ValueError:
            print('Invalid amount. Please try again.')
    elif choice == '2':
        print_report(transactions)
    elif choice == '3':
        save_to_csv(transactions)
        print('Goodbye!')
        break
    else:
        print('Invalid choice. Try again.')
```


## Project 2 — Quiz Application (Complete Step-by-Step)

### Overview
Build an interactive quiz app that loads questions from a file, presents them to the user, collects answers, and scores the result. This reinforces: file I/O, loops, conditionals, and user input.

### Step 1: Store questions in a structured format
Use a list of dictionaries or a JSON file:
```json
[
  {"question": "What is 2+2?", "options": ["3", "4", "5"], "correct": "4"},
  {"question": "What is the capital of France?", "options": ["London", "Paris", "Berlin"], "correct": "Paris"}
]
```

### Step 2: Create functions
- `load_quiz(filename)` — load questions from file
- `run_quiz(questions)` — present each question and collect answers
- `score_quiz(answers, correct)` — compare user answers to correct answers
- `print_score(score, total)` — display final result

### Step 3: Build the main flow
Load questions → present quiz → score → display result

### Step 4: Enhancements
- Add difficulty levels
- Track history
- Show explanations for incorrect answers

### Example implementation

```python
import json

def load_quiz(filename='quiz.json'):
    """Load quiz questions from JSON."""
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f'{filename} not found.')
        return []

def run_quiz(questions):
    """Present quiz and collect answers."""
    answers = []
    for i, item in enumerate(questions, 1):
        print(f'\n--- Question {i} of {len(questions)} ---')
        print(item['question'])
        for j, option in enumerate(item['options'], 1):
            print(f'{j}. {option}')
        choice = input('Your answer (1-3): ').strip()
        try:
            idx = int(choice) - 1
            if 0 <= idx < len(item['options']):
                answers.append(item['options'][idx])
            else:
                print('Invalid choice. Skipping.')
                answers.append(None)
        except ValueError:
            print('Invalid input. Skipping.')
            answers.append(None)
    return answers

def score_quiz(answers, questions):
    """Score the quiz."""
    score = 0
    for answer, question in zip(answers, questions):
        if answer == question['correct']:
            score += 1
    return score

def print_score(score, total):
    """Print final score."""
    percentage = (score / total * 100) if total > 0 else 0
    print(f'\n--- Quiz Complete ---')
    print(f'Score: {score}/{total} ({percentage:.1f}%)')
    if percentage >= 80:
        print('Excellent!')
    elif percentage >= 60:
        print('Good job!')
    else:
        print('Keep practicing!')

# Main program
questions = load_quiz()
if questions:
    answers = run_quiz(questions)
    score = score_quiz(answers, questions)
    print_score(score, len(questions))
```


## Project 3 — Todo List Manager with File Persistence

### Overview
Build a command-line todo app with persistent storage using JSON. This reinforces: file I/O, lists, dictionaries, error handling, and OOP basics.

### Step 1: Design the data model
Each todo item: `{"id": 1, "task": "Buy milk", "done": false}`. Store in a JSON file.

### Step 2: Create functions
- `load_todos(filename)` — load todos from file
- `save_todos(todos, filename)` — save todos to file
- `add_todo(todos, task)` — add new todo
- `mark_done(todos, id)` — mark todo as complete
- `remove_todo(todos, id)` — delete a todo
- `list_todos(todos)` — display all todos

### Step 3: Build the menu
Loop through: add, list, mark done, delete, save and quit.

### Example implementation

```python
import json
from pathlib import Path

class TodoApp:
    def __init__(self, filename='todos.json'):
        self.filename = filename
        self.todos = self.load()
        self.next_id = max([t['id'] for t in self.todos], default=0) + 1
    
    def load(self):
        """Load todos from file."""
        if Path(self.filename).exists():
            with open(self.filename, 'r', encoding='utf-8') as f:
                return json.load(f)
        return []
    
    def save(self):
        """Save todos to file."""
        with open(self.filename, 'w', encoding='utf-8') as f:
            json.dump(self.todos, f, indent=2)
        print('Todos saved.')
    
    def add(self, task):
        """Add a new todo."""
        self.todos.append({'id': self.next_id, 'task': task, 'done': False})
        self.next_id += 1
        print(f'Added: {task}')
    
    def list(self):
        """Display all todos."""
        if not self.todos:
            print('No todos.')
            return
        for todo in self.todos:
            status = '✓' if todo['done'] else ' '
            print(f"[{status}] {todo['id']}: {todo['task']}")
    
    def mark_done(self, todo_id):
        """Mark todo as complete."""
        for todo in self.todos:
            if todo['id'] == todo_id:
                todo['done'] = True
                print(f'Marked done: {todo["task"]}')
                return
        print('Todo not found.')
    
    def remove(self, todo_id):
        """Remove a todo."""
        self.todos = [t for t in self.todos if t['id'] != todo_id]
        print(f'Removed todo {todo_id}.')

# Main program
app = TodoApp()

while True:
    print('\n1. Add todo')
    print('2. List todos')
    print('3. Mark done')
    print('4. Remove')
    print('5. Save and quit')
    choice = input('Choose: ').strip()
    
    if choice == '1':
        task = input('Todo: ').strip()
        if task:
            app.add(task)
    elif choice == '2':
        app.list()
    elif choice == '3':
        app.list()
        try:
            todo_id = int(input('Todo ID: '))
            app.mark_done(todo_id)
        except ValueError:
            print('Invalid ID.')
    elif choice == '4':
        app.list()
        try:
            todo_id = int(input('Todo ID to remove: '))
            app.remove(todo_id)
        except ValueError:
            print('Invalid ID.')
    elif choice == '5':
        app.save()
        print('Goodbye!')
        break
    else:
        print('Invalid choice.')
```
