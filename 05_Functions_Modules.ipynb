## Writing functions

Functions package behavior into names. Keep them small and descriptive.

```python
def square(x):
    return x * x

print(square(7))

def introduce(name, city='Unknown'):
    return f'{name} lives in {city}.
'
print(introduce('Sam', city='Seattle'))
```

## Arguments and return values

Use default arguments, keyword arguments, and return multiple values with tuples.

```python
def min_max(values):
    return min(values), max(values)

low, high = min_max([4, 9, 1, 7])
print(low, high)
```

## Modules and imports

Use `import` to bring in Python libraries and local code files.

```python
import math
print('Sqrt 16:', math.sqrt(16))

import random
print('Random choice:', random.choice(['red', 'green', 'blue']))
```

## Exercise 1 — Word count function

Write a function that takes a string and returns the number of words. Use `.split()`.

```python
def count_words(text):
    return len(text.split())

print(count_words('Write a function for word count'))
```

## Exercise 2 — Keyword arguments

Define a function `describe_pet(name, species='dog')` and call it with both positional and keyword arguments.

```python
def describe_pet(name, species='dog'):
    return f'{name} is a {species}.'

print(describe_pet('Bella'))
print(describe_pet('Milo', species='cat'))
```

## Mini-Project — Conversion Module

Build a set of functions to convert between units: Celsius/Fahrenheit and kilometers/miles.

```python
def celsius_to_fahrenheit(c):
    return c * 9/5 + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

print('0C =', celsius_to_fahrenheit(0))
print('100F =', fahrenheit_to_celsius(100))
```
