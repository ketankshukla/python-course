## Numeric Expressions

Python supports basic arithmetic with `+`, `-`, `*`, `/`, `//`, `%`, and `**`. Parentheses control evaluation order.

```python
a = 7
b = 3
print('Add', a + b)
print('Subtract', a - b)
print('Multiply', a * b)
print('Divide', a / b)
print('Floor divide', a // b)
print('Remainder', a % b)
print('Power', a ** b)
print('Mixed precedence', 2 + 3 * 4)
print('With parentheses', (2 + 3) * 4)
```

## Strings: creation, indexing, and slicing

Strings are sequences. Use single or double quotes, and slice with `[start:end]`.

```python
text = 'Python Basics'
print(text)
print('First char:', text[0])
print('Slice:', text[0:6])
print('Last char:', text[-1])
print('Every second char:', text[::2])
```

## String formatting and f-strings

Formatted strings make output readable and dynamic.

```python
name = 'Alex'
age = 25
print('Name: ' + name + ', age: ' + str(age))
print(f'Name: {name}, age: {age}')
print('Upper:', name.upper())
print('Replace:', name.replace('A', '@'))
```

## Type conversion

Use `int()`, `float()`, and `str()` to convert between numeric and string types. Beware of invalid input.

```python
value = '123'
number = int(value)
print(number, type(number))
pi_text = '3.14'
pi_value = float(pi_text)
print(pi_value, type(pi_value))
print('Concatenate string:', 'value=' + str(number))
```
