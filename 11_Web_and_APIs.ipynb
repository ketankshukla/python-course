## Using `requests`

Requests is the standard way to talk to HTTP services from Python.

```python
import requests
response = requests.get('https://httpbin.org/get')
print('Status:', response.status_code)
data = response.json()
print(data['url'])
```

## Working with JSON APIs

Most APIs return JSON. Parse it and access nested fields.

```python
response = requests.get('https://api.github.com')
if response.ok:
    info = response.json()
    print('Current user url:', info['user_url'])
else:
    print('Request failed with', response.status_code)
```

## GET vs POST

GET requests fetch data; POST requests submit data. Use the `json` or `data` parameter for payloads.

```python
payload = {'name': 'Code Learner'}
response = requests.post('https://httpbin.org/post', json=payload)
print(response.json()['json'])
```

## Basic HTML parsing

Use BeautifulSoup to parse HTML and extract text or links.

```python
from bs4 import BeautifulSoup
html = '<html><body><h1>Title</h1><p>Paragraph.</p></body></html>'
soup = BeautifulSoup(html, 'html.parser')
print(soup.h1.text)
print(soup.p.text)
```

Parse a simple HTML snippet to extract text from paragraphs or headers.

```python
html = '<div><h2>Welcome</h2><p>Learn Python.</p></div>'
soup = BeautifulSoup(html, 'html.parser')
print(soup.h2.text)
print(soup.p.text)
```
